## ITERATOR 

```python
class Count_Up_To : #iterator 
    def __init__(self,max):
        self.max =max
        self.current +=1 
    def __iter__(self):
        return self
    def __next__(self):
        if self.current > self.max:
            raise StopIteration 
        else :
            val = self.current
            self.current ==  1 
            return val 

    
```

```python
counter = Count_Up_To(3)
counter
```

```python
next(counter)
counter.__next__()
```
