## ITERATOR 

Simple Iterator

```python
numbers = [1, 2, 3]

iterator = iter(numbers)

print(next(iterator))
print(next(iterator))
print(iterator.__next__())
```

```python
class Count_Up_To:
    def __init__(self, max):
        self.max = max
        self.current = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current > self.max:
            raise StopIteration
        else:
            val = self.current
            self.current += 1
            return val
```

```python
counter = Count_Up_To(3)
counter
```

```python
next(counter)
counter.__next__()
```

```python
counter.__next__()
```

3. Generator

A generator is an easier way to create iterators.

Instead of writing:

__iter__
__next__

you use:

yield

```python
def count():
    yield 1
    yield 2
    yield 3
```

```python
gen = count()

print(next(gen))
print(next(gen))
print(next(gen))
```
