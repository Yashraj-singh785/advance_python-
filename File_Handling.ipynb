## COMMON FILES EXTENSIONS IN ML :
 ### 1 . READ FILE 
data reading = database + file  
| Extension  | Meaning                    | Use                            |
| ---------- | -------------------------- | ------------------------------ |
| `.csv`     | Comma Separated Values     | Most common dataset format     |
| `.tsv`     | Tab Separated Values       | Similar to CSV                 |
| `.xlsx`    | Excel File                 | Spreadsheet datasets           |
| `.json`    | JavaScript Object Notation | APIs, configs, structured data |
| `.txt`     | Plain Text                 | Notes, NLP datasets            |
| `.parquet` | Columnar storage format    | Big data / fast ML pipelines   |
| `.txt`                                  | Raw text        |
| `.jsonl`                                | JSON Lines      |
| `.xml`                                  | Structured text |




## 2. MODEL FILE EXTENSION :

| Extension | Use                           |
| --------- | ----------------------------- |
| `.pkl`    | Pickle model storage          |
| `.joblib` | Efficient large model storage |
| `.h5`     | TensorFlow/Keras models       |
| `.onnx`   | Cross-platform ML models      |
| `.tflite` | TensorFlow Lite mobile models |
