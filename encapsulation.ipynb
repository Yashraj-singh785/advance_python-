```python
class BankAccount:

    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance

    # Deposit method
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"₹{amount} deposited successfully")
        else:
            print("Enter the amount in positive number")

    # Withdraw method
    def withdraw(self, amount):
        if amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print(f"₹{amount} withdrawn successfully")

    # Getter method
    def get_balance(self):
        print(f"Current balance: ₹{self.__balance}")
```

## Operator Overloading (Python)
 -> Giving extra meaning to standard operators like +, -, * for user-defined objects (classes).
 -> need - custom classes  like vectors .

```python
class vector :
    def __init__(self , x ,y ):
        self.x = x
        self.y = y 

v1 = vector(2, 3)
v2 = vector(4, 5)
v1+v2    ## unsupported operand type(s) for +: 'vector' and 'vector'
```

```python
class vector :
    def __init__(self , x ,y ):
        self.x = x
        self.y = y
    def __add__(self,other ) :  # other is the second object involved in the operation   ->b1 + b2  ->b1.__add__(b2)
        return vector(self.x + other.x , self.y + other.y)
    
    def show (self):
        print(self.x ,self.y)
    

v1 = vector(2, 3)
v2 = vector(4, 5)
(v1+v2).show()

    
```

note : both overriding and overloading in python is runtime not compile time. (interpreter type language )

## method overriding 
Method overriding means:

A child class changes the behavior of a method that it inherited from the parent class.

Same method name, same parameters, but different implementation.

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):
    def sound(self):
        print("Dog barks")


class Cat(Animal):
    def sound(self):
        print("Cat meows")

a1 = Dog()
a2 = Cat()

a1.sound()
a2.sound()
```

## duck typing  
-> “If it behaves like a duck, treat it like a duck.”
-> So Python doesn’t care about the object’s type — it only cares whether it has the required method or behavior.

```python
class Dog:
    def sound(self):
        print("Bark")


class Cat:
    def sound(self):
        print("Meow")


def make_sound(animal):
    animal.sound()
```

```python
make_sound(Dog())
make_sound(Cat())
```

## abstract class and methods 

```python
from abc import ABC , abstractmethod 
```

```python
class shape (ABC) :

    def area (self):
        pass 

type (shape)  # abc.ABCMeta
shape1 = shape()
```

```python
class shape () : # 

    def area (self):
        pass 
type (shape) # type
```

```python
class shape (ABC) :
    @abstractmethod
    def area (self):
        pass 

type (shape)
shape1 = shape() ## Can't instantiate abstract class shape without an implementation for abstract method 'area'
```

```python
class shape (ABC) :
    @abstractmethod
    def area (self):
        pass 

class circle (shape):
    def __init__(self , radius ):
        self.radius = radius 

circle1 = circle() ## Can't instantiate abstract class circle without an implementation for abstract method 'area
```

```python
class shape (ABC) :
    @abstractmethod
    def area (self):  ## abstract 
        pass 

class circle (shape,):
    def __init__(self , radius ):
        self.radius = radius 

    def area (self):
        return 3.14*self.radius**2
    
circle1 = circle(7) ## now no error
circle1.area()
```

```python
" Abstract + Concrete method "
class shape (ABC) :
    @abstractmethod
    def area (self):  ## abstract 
        pass 

    def concrete_method(self): # concrete 
        print(" hello I am concrete method from a Abstract class ")

class circle (shape,):
    def __init__(self , radius ):
        self.radius = radius 

    def area (self):
        return 3.14*self.radius**2
    
    def concrete_method(self): # concrete 
        print(" hello I am concrete method from a Abstract class circle  ")
    
circle1 = circle(7) ## now no error
circle1.concrete_method()
```

## Abstraction  in python 
-> what to do not how to do 
-> to hide implementation detail  

```python
class payment_method(ABC):
    @abstractmethod
    def pay (self , amount ):
        pass
    
class UPI (payment_method):
    def pay(self ,amount):
        print(f"payment RS{amount} is done using UPI ")

class Creditcard (payment_method):
    def pay(self ,amount):
        print(f"payment RS{amount} is done using Creditcard ")

def process_payment(method : payment_method , amount) :
    return method().pay(amount)

process_payment(UPI,700)
```
