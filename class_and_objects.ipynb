## pizza program 

```python
class pizza :
    def __init__(self, size , toppings ):
        self.size = size 
        self.toppings =toppings 

    def bake (self):
        print(f"your pizza is baking with the toppings{', '.join(self.toppings)} , and size of {self.size} ")

    def serve (self):
        print("your pizza is ready to serve ")
    
```

```python
req_for_pizza_by_yashraj = [[4,"large",["capsicum","olives"]],
                            [3,"medium" , ["chese","garlic"]]]
```

```python
pizza_for_yashraj = []

for req in req_for_pizza_by_yashraj:
    for quantity in range(req[0]):
        pizza_object = pizza(req[1], req[2])

        pizza_object.bake()
        pizza_object.serve()

        pizza_for_yashraj.append(pizza_object)
```


## inheritance 

```python
class parent:
    def show(self) :
        print(" this is the parent class ")

class child (parent) :
    def display(self) :
        print("this id the child class ")

c1 = child()
c1.show()   # inherited from Parent
c1.display ()   # own method of Child
```

3. Why Inheritance is Useful?
Code reusability (no need to rewrite same code)
Easier maintenance
Logical structure (hierarchy)
Extensibility (can add new features easily)

##  Types of Inheritance in Python
a. single ->one child inherits from the parent 
  multiple -> one child inherits from multiple parents 
  herarichal ->   ## one parent multiple children 
  multilevel -> chain of inheritance 
each example is give below 

```python
## single inheritance 

class Animal :
    def show(self):
        print("animal make sounds ")

class cat (Animal) :
    def meows(self):
        print("cat say i meows ")

cat1 = cat()
cat1. show ()
cat1.meows()
```

```python
 ## multiple inheritance ::
class father :
    def skill1(self):
        print("Indian army ")

class mother :
    def skill2(self) :
        print("teacher")

class yashraj(father,mother) :
    def skill3(self):
        print(" nothing as great as my father and mother ")

yash  = yashraj()
yash.skill1()
yash.skill2()
yash.skill3()
```

```python
## multilevel inheritance 

class Grandparent:
    def show1(self):
        print("Grandparent class")

class Parent(Grandparent):
    def show2(self):
        print("Parent class")

class Child(Parent):
    def show3(self):
        print("Child class")

c = Child()
c.show1()
c.show2()
c.show3()
```

```python
## herarichal inheritance 
class Vehicle:
    def info(self):
        print("This is a vehicle")

class Car(Vehicle):
    pass

class Bike(Vehicle):
    pass

c = Car()
b = Bike()

c.info()
b.info()
```

```python
## super keyword example -> used to call  parent clss methods 
class parent :
    def show (self):
        print("parent class method ")

class child (parent):
    def show (self ):
        super().show()
        print("this is child method ")
    
child2= child ()
child2.show()
```

```python
## real world exaples :

class Employee:
    def __init__(self, name):
        self.name = name

    def show(self):
        print("Employee Name:", self.name)

class Manager(Employee):
    def __init__(self, name, department):
        super().__init__(name)
        self.department = department

    def display(self):
        print("Department:", self.department)

m = Manager("Amit", "IT")
m.show()
m.display()
```
