#  sql , sqlite , sqlite 3
SQL → A language used to communicate with and manage databases.
SQLite → A lightweight, file-based database system that uses SQL.
sqlite3 → Python’s built-in module to interact with SQLite databases using Python code.

```python
# 1.Import sqlite3 module :
import sqlite3
```

```python
# 2. Create Database Connection :
conn = sqlite3.connect("students.db") # If file doesn't exist -> SQLite creates it automatically.
```

```python
# 5. Create Cursor -> Cursor executes SQL commands.

cursor = conn.cursor()
```

```python
# 6. Create Table
cursor.execute("""
CREATE TABLE students (
    id INTEGER PRIMARY KEY,
    name TEXT,
    age INTEGER                          
)
""")
               
               
```

```python
## Inserting the data :

cursor.execute("""
INSERT INTO students VALUES
            (1, 'yahsraj', 19),
            (2, 'Aman' , 17 ),
            (3, 'Annu'  , 20 ),
            (4, 'Imran' , 18)
                 
                 """)
```

```python
## Quering the data :
result = cursor.execute("SELECT * from students")
for row in result :
    print(row)
```

```python
##  data should eb sorted based upon the experince:
result = cursor.execute("SELECT * from students ORDER BY age")
for row in result :
    print (row )
```

## Reading the database and writing to a csv/excel file

```python
result = conn.execute("SELECT * from students " )
students =[]
for row in result :
    students.append(row)
print(students)
```

```python
import pandas as pd 
df = pd.DataFrame(students , columns =["id" , "name" , "age" ] ) # converting into file

df.head()
```

```python
## saving to csv form :
df.to_csv("students_csv.csv" , index =False)
```

```python
conn.commit() # Without commit → changes may not save.
```

```python
 # 8.Close database
conn.close()
```
