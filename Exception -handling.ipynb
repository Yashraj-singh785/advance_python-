## Exception handling

->An exception is an event/error that occurs during program execution(runtime) and interrupts the normal flow of the program.

Instead of crashing silently, Python creates an exception object.

```python
x= 10/0 #ZeroDivisionError: division by zero
```

 ## Why Exceptions are Important

1.Without exception handling:

Program terminates abruptly
Bad user experience
Critical systems may fail

2.With exception handling:

Program continues safely
Errors handled gracefully
Logging/debugging becomes easier

Errors	                            Exceptions
Serious problem	                     Can be handled
Mostly syntax/system related	     Runtime issue
Program usually stops	             We can recover
Hard/impossible to handle	         Handled using try-except

```python
# finally ->must execute block (excep.occur ir not ) , use ->closing resources
try:
    result= 10/0
except Exception as  e:
    print(" problem")
finally:
    print("I will always execute ")
```
