## geter and setter  with application in ai ml 

```python
import numpy as np 
```

```python
class Dataloader :
    def __init__(self):
        self._data = None
    
    def get_data(self):
        return self._data
    
    def set_data(self ,arr):
        if not isinstance (arr,np.ndarray) :
            raise TypeError ("data should be in numpy array ")
        
        if arr.ndim != 2 :
            raise ValueError("Input data must be in 2-d ")
        print(f'shape {arr.shape}')
        self._data = arr 
```

```python
loader =Dataloader()
data =[1,2,34]
loader.set_data(data)
```

```python
data = np.array([1,2,3,4])
loader.set_data(data)
```

@ property  and  property ()

```python
class Dataloader :
    def __init__(self ) :
        self._data = None
    
    def get_data(self ):
        return self._data
    
    def set_data (self , arr ):
        if not isinstance(arr,np.adarray)
            raise TypeError("array must be numpy ")

        if arr.ndim !=2 :
            raise ValueError("array should be 2d ")
        print (f' shape : {arr.shape}')
        self._data = arr 
    
    data = property (get_data  , set_data)
```

```python
loader =Dataloader()

value =np.ones((3,4))
loader.set_data(value )
print(value )
```

```python
loader.data = np.ones((4,3))
loader.data ## why because of property ()
```

@ property  most appreciated approach 

```python
class Dataloader :
    def __init__(self ) :
        self._data = None
    
    @ property 
    def data(self ):
        return self._data
    
    @data.setter 
    def data(self , arr ):
        if not isinstance(arr,np.ndarray):
            raise TypeError("array must be numpy ")

        if arr.ndim !=2 :
            raise ValueError("array should be 2d ")
        print (f' shape : {arr.shape}')
        self._data = arr 
    
   
```

```python
loader = Dataloader()
loader.data = np.ones((3,4))
loader.data
```
